## recommendations.csv

* Filtrado de de fichero de recomendaciones (~2GB) a Elden Ring y GTA IV
* Mi PC no tiene memoria suficiente para hacerlo de una tacada con `df.read_csv()`, tengo que hacerlo por chunks

```python
import pandas as pd

target_games = [1245620]
filtered_chunks = []
chunk_size = 500_000 # 500k registros

with pd.read_csv("recommendations.csv", chunksize=chunk_size) as reader:
    for i,chunk in enumerate(reader):
        print(f"Chunk {i}")
        # chunk is the goddamn dataframe
        filtered = chunk[chunk["app_id"].isin(target_games)]
        if not filtered.empty:
            filtered_chunks.append(filtered)

df = pd.concat(filtered_chunks,ignore_index=True)
```
